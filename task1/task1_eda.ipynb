## 3. Dataset Description
The Titanic passenger dataset contains information about passengers such as:
- `PassengerId`: unique passenger number.
- `Survived`: whether the passenger survived (0 = Did not survive, 1 = Survived).
- `Pclass`: passenger class (1 = First class, 2 = Second class, 3 = Third class).
- `Name`: passenger name.
- `Sex`: passenger gender.
- `Age`: passenger age.
- `SibSp`: number of siblings/spouses travelling with the passenger.
- `Parch`: number of parents/children travelling with the passenger.
- `Ticket`: ticket number.
- `Fare`: amount paid for ticket.
- `Cabin`: cabin number.
- `Embarked`: port from which passenger boarded (C = Cherbourg, Q = Queenstown, S = Southampton).

## Part I - Data Cleaning
### Task 16: Remove Duplicate Rows

```python
df = df.drop_duplicates().copy()
print("Remaining rows after drop_duplicates:", len(df))
```

### Task 17: Handle Missing Age Values

```python
median_age = df["Age"].median()
print("Median Age:", median_age)
df["Age"] = df["Age"].fillna(median_age)
```

### Task 18: Handle Missing Embarked Values

```python
mode_embarked = df["Embarked"].mode()[0]
print("Mode Embarked:", mode_embarked)
df["Embarked"] = df["Embarked"].fillna(mode_embarked)
```

### Task 19: Verify Missing Values After Cleaning

```python
df.isnull().sum()
```

#### Student Questions Answers (Task 19):
1. **Does Age still contain missing values?** No (0 missing).
2. **Does Embarked still contain missing values?** No (0 missing).
3. **Which column still contains a large number of missing values?** `Cabin` (687 missing, kept unchanged per instructions).

### Question 4: How were missing values in Age handled? Explain why the median was used.
Missing values in `Age` were imputed using the median age (**28.0 years**). The median was chosen because age distributions can be skewed and contain outliers; the median is a robust measure of central tendency that is resistant to extreme values unlike the mean.

### Question 5: How were missing values in Embarked handled? Explain why the mode is suitable for this feature.
Missing values in `Embarked` were imputed using the mode (**'S'**, Southampton). The mode is the most suitable approach because `Embarked` is a nominal categorical variable with only two missing values (0.22%), and Southampton accounted for over 72% of all boarding passengers.

### Question 6: Were duplicate rows found in the dataset?
No duplicate rows were found in the dataset (`df.duplicated().sum() == 0`).

### Question 7: What is the overall survival rate?
The overall survival rate was **38.38%** (342 out of 891 passengers survived).

### Question 8: Which sex had the higher survival rate?

### Question 9: Which passenger class had the highest survival rate?
**1st Class** had the highest survival rate (**62.96%**), followed by 2nd Class (**47.28%**) and 3rd Class (**24.24%**).

## Final Conclusion
The Titanic dataset contains 891 passenger records across 12 features covering demographic, socioeconomic, and travel characteristics. Significant missing values were identified in `Cabin` (77.10%), `Age` (19.87%), and `Embarked` (0.22%), which were addressed via domain-appropriate strategies: leaving Cabin untreated, imputing Age with the median (28.0) to resist outlier distortion, and imputing Embarked with its mode ('S'). Survival analysis revealed two primary drivers of passenger outcome: gender (females survived at 74.20% vs. 18.89% for males) and socioeconomic status (1st Class survived at 62.96% vs. 24.24% for 3rd Class). Numerical distributions highlighted massive right-skewness and genuine luxury outliers in `Fare` alongside a symmetric age distribution centered around young adults. Performing this EDA provided foundational insights into feature importance, data quality limitations, and structural relationships, directly preparing the dataset for effective predictive modeling.
